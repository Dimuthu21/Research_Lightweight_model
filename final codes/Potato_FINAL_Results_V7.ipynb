```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
from torchvision import transforms, datasets, models
from sklearn.metrics import precision_recall_fscore_support, confusion_matrix
import numpy as np
import csv
from tqdm import tqdm
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import argparse
import traceback
import timm
import random
from pathlib import Path
import logging
from PIL import Image
import torch.nn.functional as F
import copy
import json


# Custom ImageFolder to handle missing files
class SafeImageFolder(datasets.ImageFolder):
    def __init__(self, root, transform=None):
        super().__init__(root, transform=transform)
        self.invalid_files = []

    def __getitem__(self, index):
        path, target = self.samples[index]
        try:
            sample = self.loader(path)
            if self.transform is not None:
                sample = self.transform(sample)
            return sample, target
        except Exception as e:
            print(f"Error loading {path}: {e}")
            self.invalid_files.append(path)
            dummy = torch.zeros(3, 224, 224) if self.transform is None else self.transform(Image.new('RGB', (224, 224)))
            return dummy, -1  # -1 indicates invalid sample

    def __len__(self):
        return len(self.samples)

# Early stopping class
class EarlyStopping:
    def __init__(self, patience=10, delta=0.001, verbose=True, path='checkpoint.pt'):
        self.patience = patience
        self.delta = delta
        self.verbose = verbose
        self.path = path
        self.best_score = None
        self.early_stop = False
        self.counter = 0
        self.best_f1 = 0.0

    def __call__(self, val_f1, model):

        if self.best_score is None:
            self.best_score = val_f1
            self.save_checkpoint(val_f1, model)

        elif val_f1 < self.best_score + self.delta:
            self.counter += 1

            if self.verbose:
                print(
                    f'EarlyStopping counter: {self.counter} out of {self.patience}'
                )

            if self.counter >= self.patience:
                self.early_stop = True

        else:
            self.best_score = val_f1
            self.save_checkpoint(val_f1, model)
            self.counter = 0

    def save_checkpoint(self, val_f1, model):

        if self.verbose:
            print(
                f'Validation F1 improved ({self.best_f1:.4f} --> {val_f1:.4f}). Saving model ...'
            )

        torch.save(model.state_dict(), self.path)
        self.best_f1 = val_f1

class StudentNet(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.base = models.efficientnet_b0(weights=models.EfficientNet_B0_Weights.IMAGENET1K_V1)
        in_ch = self.base.classifier[1].in_features
        self.base.classifier[1] = nn.Linear(in_ch, num_classes)
        self.feature_adapter = nn.Linear(1280, 768)

    def forward(self, x, return_features=False):
        features = self.base.features(x)
        features = self.base.avgpool(features)
        features = torch.flatten(features, 1)
        logits = self.base.classifier(features)
        if return_features:
            return logits, features
        return logits

# Validate dataset and list files
def validate_dataset(data_dir, class_names):
    print("Validating dataset...")
    total_images = 0
    if not os.path.exists(data_dir):
        raise FileNotFoundError(f"Directory {data_dir} does not exist")
    for class_name in class_names:
        class_path = os.path.join(data_dir, class_name)
        if not os.path.exists(class_path):
            print(f"Warning: Class directory {class_path} does not exist")
            continue
        files = [f for f in os.listdir(class_path) if f.lower().endswith(('.jpg', '.png', '.jpeg'))]
        print(f"{class_name}: {len(files)} images found")
        total_images += len(files)
        for f in files:
            file_path = os.path.join(class_path, f)
            if not os.path.isfile(file_path):
                print(f"Error: File {file_path} is inaccessible")
            try:
                Image.open(file_path).verify()
            except Exception as e:
                print(f"Error: Corrupted image {file_path}: {e}")
    if total_images == 0:
        raise ValueError("No valid images found in dataset")
    print(f"Dataset validation complete: {total_images} images found.")

# MixUp helper functions
def mixup_data(x, y, alpha=1.0):
    if alpha > 0:
        lam = np.random.beta(alpha, alpha)
    else:
        lam = 1
    batch_size = x.size()[0]
    index = torch.randperm(batch_size).to(x.device)
    mixed_x = lam * x + (1 - lam) * x[index, :]
    y_a, y_b = y, y[index]
    return mixed_x, y_a, y_b, lam

def mixup_criterion(criterion, pred, y_a, y_b, lam):
    return lam * criterion(pred, y_a) + (1 - lam) * criterion(pred, y_b)


# Training function with metrics and early stopping
def train_model(model, train_loader, val_loader, criterion, optimizer, scheduler, num_epochs, device, early_stopping, variant, output_dir):
    model.to(device)
    train_losses, val_losses = [], []
    train_accs, val_accs = [], []
    train_precisions, train_recalls, train_f1s = [], [], []
    val_precisions, val_recalls, val_f1s = [], [], []
    epoch_summaries = []

    for epoch in range(num_epochs):
        model.train()
        running_loss = 0.0
        correct = 0
        total = 0
        all_preds, all_labels = [], []

        for images, labels in tqdm(train_loader, desc=f"Epoch {epoch+1}/{num_epochs}", leave=False):
            valid_mask = labels != -1
            if not valid_mask.any():
                continue
            images, labels = images[valid_mask].to(device), labels[valid_mask].to(device)

            # === Apply MixUp ===
            images, y_a, y_b, lam = mixup_data(images, labels, alpha=0.4)

            optimizer.zero_grad()
            outputs = model(images)
            loss = mixup_criterion(criterion, outputs, y_a, y_b, lam)
            loss.backward()
            optimizer.step()


            running_loss += loss.item() * images.size(0)
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
            all_preds.extend(predicted.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

        train_loss = running_loss / total if total > 0 else float('inf')
        train_acc = 100 * correct / total if total > 0 else 0
        train_precision, train_recall, train_f1, _ = precision_recall_fscore_support(
            all_labels, all_preds, average='weighted', zero_division=0
        )

        train_losses.append(train_loss)
        train_accs.append(train_acc)
        train_precisions.append(train_precision)
        train_recalls.append(train_recall)
        train_f1s.append(train_f1)

        model.eval()
        val_loss = 0.0
        val_correct = 0
        val_total = 0
        val_preds, val_labels = [], []

        with torch.no_grad():
            for images, labels in val_loader:
                valid_mask = labels != -1
                if not valid_mask.any():
                    continue
                images, labels = images[valid_mask].to(device), labels[valid_mask].to(device)
                outputs = model(images)
                loss = criterion(outputs, labels)
                val_loss += loss.item() * images.size(0)
                _, predicted = torch.max(outputs, 1)
                val_total += labels.size(0)
                val_correct += (predicted == labels).sum().item()
                val_preds.extend(predicted.cpu().numpy())
                val_labels.extend(labels.cpu().numpy())

        val_loss = val_loss / val_total if val_total > 0 else float('inf')
        val_acc = 100 * val_correct / val_total if val_total > 0 else 0
        val_precision, val_recall, val_f1, _ = precision_recall_fscore_support(
            val_labels, val_preds, average='weighted', zero_division=0
        )

        val_losses.append(val_loss)
        val_accs.append(val_acc)
        val_precisions.append(val_precision)
        val_recalls.append(val_recall)
        val_f1s.append(val_f1)

        # Save epoch summary
        epoch_summary = {
            'epoch': epoch + 1,
            'train_loss': train_loss,
            'train_accuracy': train_acc,
            'train_precision': train_precision,
            'train_recall': train_recall,
            'train_f1': train_f1,
            'val_loss': val_loss,
            'val_accuracy': val_acc,
            'val_precision': val_precision,
            'val_recall': val_recall,
            'val_f1': val_f1
        }
        epoch_summaries.append(epoch_summary)
        print(f"Epoch {epoch + 1}: "
              f"Train Loss: {train_loss:.4f}, Train Acc: {train_acc:.2f}%, "
              f"Train Precision: {train_precision:.4f}, Train Recall: {train_recall:.4f}, Train F1: {train_f1:.4f}, "
              f"Val Loss: {val_loss:.4f}, Val Acc: {val_acc:.2f}%, "
              f"Val Precision: {val_precision:.4f}, Val Recall: {val_recall:.4f}, Val F1: {val_f1:.4f}, "
              f"Learning Rate: {optimizer.param_groups[0]['lr']:.6f}")
        # Early stopping check
        early_stopping(val_f1, model)
        if early_stopping.early_stop:
            print("Early stopping triggered")
            break
        # Step the scheduler
        scheduler.step()
    # Save epoch summaries to CSV
    epoch_df = pd.DataFrame(epoch_summaries)
    csv_path = os.path.join(output_dir, f'epoch_summary_{variant}.csv')
    epoch_df.to_csv(csv_path, index=False)
    print(f"Saved epoch summaries to {csv_path}")
    # Load best model weights
    try:
        model.load_state_dict(torch.load(early_stopping.path, weights_only=True))
    except Exception as e:
        print(f"Error loading best model weights: {e}")
        raise
    # Plot metrics
    plot_metrics(train_losses, val_losses, train_accs, val_accs,
                 train_precisions, train_recalls, train_f1s,
                 val_precisions, val_recalls, val_f1s,
                 val_labels, val_preds, variant, output_dir)
    # Print training accuracies from CSV
    print("\nTraining Accuracies per Epoch:")
    try:
        print(epoch_df[['epoch', 'train_accuracy']])
    except Exception as e:
        print(f"Error reading training accuracies from DataFrame: {e}")
    return model

def train_student_model(model, teacher_model, train_loader, val_loader, criterion, optimizer, scheduler, num_epochs, device, early_stopping, variant, output_dir):
    model.to(device)
    teacher_model.to(device)
    train_losses, val_losses = [], []
    train_accs, val_accs = [], []
    train_precisions, train_recalls, train_f1s = [], [], []
    val_precisions, val_recalls, val_f1s = [], [], []
    epoch_summaries = []

    alpha = 0.4
    T = 4.0
    beta = 0.0

    for epoch in tqdm(range(num_epochs), desc="Epochs"):
        model.train()
        running_loss = 0.0
        correct = 0
        total = 0
        all_preds, all_labels = [], []

        for images, labels in train_loader:
            valid_mask = labels != -1
            if not valid_mask.any():
                continue
            images, labels = images[valid_mask].to(device), labels[valid_mask].to(device)

            # Step 1: Teacher gets CLEAN images → soft labels
            with torch.no_grad():
                teacher_outputs = teacher_model(images)

            # Step 2: Apply MixUp for the student
            mixed_images, y_a, y_b, lam = mixup_data(images, labels, alpha=0.4)

            # Step 3: student sees mixed images
            outputs = model(mixed_images, return_features=False)

            # Logit distillation loss (KL divergence)
            kd_loss = F.kl_div(F.log_softmax(outputs / T, dim=1), F.softmax(teacher_outputs / T, dim=1), reduction='batchmean') * (T * T)

            # CE loss on mixed labels
            ce_loss = lam * criterion(outputs, y_a) + (1 - lam) * criterion(outputs, y_b)

            loss = alpha * kd_loss + (1 - alpha) * ce_loss

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            running_loss += loss.item() * images.size(0)
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
            all_preds.extend(predicted.cpu().numpy())
            all_labels.extend(labels.cpu().numpy())

        train_loss = running_loss / total if total > 0 else float('inf')
        train_acc = 100 * correct / total if total > 0 else 0
        train_precision, train_recall, train_f1, _ = precision_recall_fscore_support(
            all_labels, all_preds, average='weighted', zero_division=0
        )

        train_losses.append(train_loss)
        train_accs.append(train_acc)
        train_precisions.append(train_precision)
        train_recalls.append(train_recall)
        train_f1s.append(train_f1)

        model.eval()
        val_loss = 0.0
        val_correct = 0
        val_total = 0
        val_preds, val_labels = [], []

        with torch.no_grad():
            for images, labels in val_loader:
                valid_mask = labels != -1
                if not valid_mask.any():
                    continue
                images, labels = images[valid_mask].to(device), labels[valid_mask].to(device)
                outputs = model(images)
                loss = criterion(outputs, labels)
                val_loss += loss.item() * images.size(0)
                _, predicted = torch.max(outputs, 1)
                val_total += labels.size(0)
                val_correct += (predicted == labels).sum().item()
                val_preds.extend(predicted.cpu().numpy())
                val_labels.extend(labels.cpu().numpy())

        val_loss = val_loss / val_total if val_total > 0 else float('inf')
        val_acc = 100 * val_correct / val_total if val_total > 0 else 0
        val_precision, val_recall, val_f1, _ = precision_recall_fscore_support(
            val_labels, val_preds, average='weighted', zero_division=0
        )

        val_losses.append(val_loss)
        val_accs.append(val_acc)
        val_precisions.append(val_precision)
        val_recalls.append(val_recall)
        val_f1s.append(val_f1)

        epoch_summary = {
            'epoch': epoch + 1,
            'train_loss': train_loss,
            'train_accuracy': train_acc,
            'train_precision': train_precision,
            'train_recall': train_recall,
            'train_f1': train_f1,
            'val_loss': val_loss,
            'val_accuracy': val_acc,
            'val_precision': val_precision,
            'val_recall': val_recall,
            'val_f1': val_f1
        }
        epoch_summaries.append(epoch_summary)

        print(f"Epoch {epoch + 1}: Train Loss: {train_loss:.4f}, Train Acc: {train_acc:.2f}%, Val Loss: {val_loss:.4f}, Val Acc: {val_acc:.2f}%")

        early_stopping(val_f1, model)
        if early_stopping.early_stop:
            print("Early stopping triggered")
            break
        scheduler.step()

    epoch_df = pd.DataFrame(epoch_summaries)
    csv_path = os.path.join(output_dir, f'student_epoch_summary_{variant}.csv')
    epoch_df.to_csv(csv_path, index=False)
    print(f"Saved epoch summaries to {csv_path}")

    model.load_state_dict(torch.load(early_stopping.path, weights_only=True))
    return model, epoch_df, (train_losses, val_losses, train_accs, val_accs,
                             train_precisions, train_recalls, train_f1s,
                             val_precisions, val_recalls, val_f1s,
                             val_labels, val_preds)

# ======================================================
# QAT Fine-tuning Function (AMP-based)
# ======================================================

def train_qat_student_model(student_ckpt_path, teacher_model, train_loader, val_loader, device,
                            num_epochs=30, lr=1e-5, weight_decay=0.01, patience=12,
                            variant="qat_student", output_dir="./qat"):

    import copy
    os.makedirs(output_dir, exist_ok=True)

    # --- Load trained student checkpoint ---
    student_model = StudentNet(num_classes=len(train_loader.dataset.dataset.classes)).to(device)
    state_dict = torch.load(student_ckpt_path, map_location=device)
    student_model.load_state_dict(state_dict)
    student_model.eval()

    # --- Copy for QAT training ---
    model_qat = copy.deepcopy(student_model).to(device)
    optimizer = torch.optim.AdamW(model_qat.parameters(), lr=lr, weight_decay=weight_decay)
    scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=num_epochs)
    scaler = torch.cuda.amp.GradScaler(enabled=(device.type=="cuda"))

    best_val_acc = 0.0
    patience_counter = 0
    best_model_path = os.path.join(output_dir, f"{variant}_best.pth")
    alpha = 0.4
    T = 4.0
    criterion = nn.CrossEntropyLoss(label_smoothing=0.1)

    # === Tracking lists ===
    train_losses, val_losses = [], []
    train_accs, val_accs = [], []

    # --- Training Loop ---
    for epoch in range(num_epochs):
        model_qat.train()
        train_loss, seen = 0.0, 0
        train_correct, train_total = 0, 0

        for images, labels in train_loader:
            valid_mask = labels != -1
            if not valid_mask.any(): continue
            images, labels = images[valid_mask].to(device), labels[valid_mask].to(device)

            # --- MixUp augmentation ---
            # Step 1: Teacher on clean images
            with torch.no_grad():
                teacher_logits = teacher_model(images)

            # Step 2: MixUp for student
            mixed_images, y_a, y_b, lam = mixup_data(images, labels, alpha=0.4)

            optimizer.zero_grad(set_to_none=True)
            with torch.cuda.amp.autocast(enabled=(device.type=="cuda")):
                student_logits = model_qat(mixed_images, return_features=False)

                kd_loss = F.kl_div(F.log_softmax(student_logits / T, dim=1),
                                   F.softmax(teacher_logits / T, dim=1),
                                   reduction='batchmean') * (T * T)

                ce_loss = lam * criterion(student_logits, y_a) + (1 - lam) * criterion(student_logits, y_b)
                loss = alpha * kd_loss + (1 - alpha) * ce_loss

            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()

            train_loss += loss.item() * images.size(0)
            seen += images.size(0)

            # track training accuracy
            preds = student_logits.argmax(1)
            train_correct += (preds == labels).sum().item()
            train_total += labels.size(0)

        train_loss /= max(seen, 1)
        train_acc = 100.0 * train_correct / max(train_total, 1)

        # --- Validation ---
        model_qat.eval()
        val_loss, val_correct, val_total = 0.0, 0, 0

        with torch.no_grad():
            for images, labels in val_loader:
                valid_mask = labels != -1
                if not valid_mask.any(): continue
                images, labels = images[valid_mask].to(device), labels[valid_mask].to(device)

                with torch.cuda.amp.autocast(enabled=(device.type=="cuda")):
                    teacher_logits = teacher_model(images)
                    student_logits = model_qat(images, return_features=False)

                    # --- KD loss again ---
                    kd_loss = F.kl_div(F.log_softmax(student_logits / T, dim=1),
                                       F.softmax(teacher_logits / T, dim=1),
                                       reduction='batchmean') * (T * T)
                    ce_loss = criterion(student_logits, labels)
                    vloss = alpha * kd_loss + (1 - alpha) * ce_loss

                val_loss += vloss.item() * images.size(0)
                preds = student_logits.argmax(1)
                val_correct += (preds == labels).sum().item()
                val_total += labels.size(0)

        val_loss /= max(val_total, 1)
        val_acc = 100.0 * val_correct / max(val_total, 1)

        # store metrics
        train_losses.append(train_loss)
        val_losses.append(val_loss)
        train_accs.append(train_acc)
        val_accs.append(val_acc)

        print(f"[QAT][Epoch {epoch+1}/{num_epochs}] "
              f"Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.2f}% | "
              f"Val Loss: {val_loss:.4f} | Val Acc: {val_acc:.2f}%")

        # --- Early stopping ---
        if val_acc > best_val_acc:
            best_val_acc = val_acc
            patience_counter = 0
            torch.save(model_qat.state_dict(), best_model_path)
        else:
            patience_counter += 1
            if patience_counter >= patience:
                print("QAT early stopping triggered.")
                break

        scheduler.step()

    # --- Load best QAT checkpoint ---
    best_model = StudentNet(num_classes=len(train_loader.dataset.dataset.classes)).to(device)
    best_model.load_state_dict(torch.load(best_model_path, map_location=device))
    best_model.eval()

    fp16_model = copy.deepcopy(best_model).half()
    fp16_path = best_model_path.replace("_best.pth", "_best_fp16.pth")

    # Save clean QAT checkpoints
    _num_classes = len(train_loader.dataset.dataset.classes)
    _device      = next(best_model.parameters()).device
    _clean_fp32  = best_model_path.replace("_best.pth", "_clean_fp32.pt")
    _clean_fp16  = best_model_path.replace("_best.pth", "_clean_fp16.pt")
    print("Saving clean QAT student model...")
    save_clean_student(best_model, _clean_fp32, _clean_fp16, _num_classes, _device)

    # Return with tracked metrics
    return best_model, fp16_model, best_model_path, fp16_path, (train_losses, val_losses, train_accs, val_accs)


# Plotting function
def plot_metrics(train_losses, val_losses, train_accs, val_accs,
                 train_precisions, train_recalls, train_f1s,
                 val_precisions, val_recalls, val_f1s,
                 val_labels, val_preds, variant, output_dir):
    epochs = range(1, len(train_losses) + 1)

    plt.figure(figsize=(10, 4))
    plt.plot(epochs, train_losses, 'b-', label='Train Loss')
    plt.plot(epochs, val_losses, 'r-', label='Val Loss')
    plt.title('Training and Validation Loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    plt.savefig(os.path.join(output_dir, f'loss_plot_{variant}.png'))
    plt.show()
    plt.close()

    plt.figure(figsize=(10, 4))
    plt.plot(epochs, train_accs, 'b-', label='Train Accuracy')
    plt.plot(epochs, val_accs, 'r-', label='Val Accuracy')
    plt.title('Training and Validation Accuracy')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy (%)')
    plt.legend()
    plt.grid(True)
    plt.savefig(os.path.join(output_dir, f'accuracy_plot_{variant}.png'))
    plt.show()
    plt.close()

    plt.figure(figsize=(10, 4))
    plt.plot(epochs, val_precisions, 'b-', label='Val Precision')
    plt.plot(epochs, val_recalls, 'r-', label='Val Recall')
    plt.plot(epochs, val_f1s, 'g-', label='Val F1')
    plt.title('Validation Precision, Recall, and F1-Score')
    plt.xlabel('Epochs')
    plt.ylabel('Score')
    plt.legend()
    plt.grid(True)
    plt.savefig(os.path.join(output_dir, f'prf1_plot_{variant}.png'))
    plt.show()
    plt.close()

    cm = confusion_matrix(val_labels, val_preds)
    plt.figure(figsize=(12, 10))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                xticklabels=class_names, yticklabels=class_names)
    plt.title('Confusion Matrix')
    plt.xlabel('Predicted')
    plt.ylabel('True')
    plt.savefig(os.path.join(output_dir, f'confusion_matrix_{variant}.png'))
    plt.show()
    plt.close()

def plot_qat_metrics(train_losses, val_losses, train_accs, val_accs, variant, output_dir):
    epochs = range(1, len(train_losses) + 1)

    plt.figure(figsize=(10, 4))
    plt.plot(epochs, train_losses, 'b-', label='Train Loss')
    plt.plot(epochs, val_losses, 'r-', label='Val Loss')
    plt.title('QAT Training and Validation Loss')
    plt.xlabel('Epochs')
    plt.ylabel('Loss')
    plt.legend()
    plt.grid(True)
    plt.savefig(os.path.join(output_dir, f'qat_loss_plot_{variant}.png'))
    plt.show()
    plt.close()

    plt.figure(figsize=(10, 4))
    plt.plot(epochs, train_accs, 'b-', label='Train Accuracy')
    plt.plot(epochs, val_accs, 'r-', label='Val Accuracy')
    plt.title('QAT Training and Validation Accuracy')
    plt.xlabel('Epochs')
    plt.ylabel('Accuracy (%)')
    plt.legend()
    plt.grid(True)
    plt.savefig(os.path.join(output_dir, f'qat_accuracy_plot_{variant}.png'))
    plt.show()
    plt.close()


def get_layer_wise_params(model, base_lr=1e-4, decay=0.65):
    param_groups = []
    # Classification head — highest LR
    param_groups.append({"params": model.head.parameters(), "lr": base_lr})
    # Transformer blocks — decay LR for earlier blocks
    blocks = list(model.blocks)
    for i, block in enumerate(reversed(blocks)):
        layer_lr = base_lr * (decay ** (i + 1))
        param_groups.append({"params": block.parameters(), "lr": layer_lr})
    # Patch embedding — very low LR (already well-pretrained)
    param_groups.append({
        "params": model.patch_embed.parameters(),
        "lr": base_lr * (decay ** (len(blocks) + 1))
    })
    return param_groups

def save_clean_student(student_model, save_path_fp32, save_path_fp16, num_classes, device):
    clean = models.efficientnet_b0(weights=None)
    clean.classifier[1] = nn.Linear(clean.classifier[1].in_features, num_classes)

    student_state = {k: v.cpu() for k, v in student_model.state_dict().items()}
    clean_state   = {k[len('base.'):]: v
                     for k, v in student_state.items()
                     if k.startswith('base.')}
    clean.load_state_dict(clean_state)
    clean = clean.to(device)
    clean.eval()

    # Save FP32 clean checkpoint
    torch.save(clean.state_dict(), save_path_fp32)
    fp32_mb = sum(p.numel() for p in clean.parameters()) * 4 / 1024 / 1024
    print(f"  [Clean FP32] Saved → {save_path_fp32}  ({fp32_mb:.2f} MB)")

    # Save FP16 clean checkpoint
    clean_fp16 = None
    if save_path_fp16 is not None:
        clean_fp16 = copy.deepcopy(clean).half()
        torch.save(clean_fp16.state_dict(), save_path_fp16)
        fp16_mb = sum(p.numel() for p in clean_fp16.parameters()) * 2 / 1024 / 1024
        print(f"  [Clean FP16] Saved → {save_path_fp16}  ({fp16_mb:.2f} MB)")

    return clean, clean_fp16

# Main execution
def main(args):
    output_dir = "/content/drive/MyDrive/Research/Model_save/V7"
    os.makedirs(output_dir, exist_ok=True)
    print(f"Created output directory: {output_dir}")

    # ==== setup ====
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    batch_size = args.batch_size
    num_epochs = args.epochs
    learning_rate = args.learning_rate
    data_dir = args.data_dir

    # Define transforms
    train_transform = transforms.Compose([
        transforms.RandomResizedCrop(224, scale=(0.6, 1.0)),
        transforms.RandomHorizontalFlip(),
        transforms.RandomVerticalFlip(),
        transforms.RandomRotation(degrees=30),
        transforms.ColorJitter(brightness=0.4, contrast=0.4, saturation=0.4, hue=0.1),
        transforms.GaussianBlur(kernel_size=5, sigma=(0.1, 2.0)),
        transforms.ToTensor(),
        transforms.RandomErasing(p=0.25, scale=(0.02, 0.2), ratio=(0.3, 3.3)),
        transforms.Normalize(mean=[0.485, 0.456, 0.406],
                            std=[0.229, 0.224, 0.225])
    ])

    val_transform = transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.CenterCrop(224),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406],
                            std=[0.229, 0.224, 0.225])
    ])

    # Load dataset
    dataset = SafeImageFolder(data_dir, transform=val_transform)
    global class_names
    class_names = dataset.classes
    print(f"Classes detected: {class_names}")

    # Validate dataset
    validate_dataset(data_dir, class_names)

    num_classes = len(dataset.classes)

    # ===============================
    # 5-Fold Cross Validation Section
    # ===============================
    from sklearn.model_selection import StratifiedKFold
    from torch.utils.data import Subset, DataLoader
    from sklearn.metrics import accuracy_score, precision_recall_fscore_support

    targets = [s[1] for s in dataset.samples]
    skf = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

    fold_accuracies, fold_precisions, fold_recalls, fold_f1s = [], [], [], []

    def evaluate_on_loader(model, data_loader, device):
        model.eval()
        all_preds, all_labels = [], []
        with torch.no_grad():
            for images, labels in data_loader:
                valid_mask = labels != -1
                if not valid_mask.any():
                    continue
                images, labels = images[valid_mask].to(device), labels[valid_mask].to(device)
                outputs = model(images)
                preds = torch.argmax(outputs, dim=1)
                all_preds.extend(preds.cpu().numpy())
                all_labels.extend(labels.cpu().numpy())
        acc = accuracy_score(all_labels, all_preds)
        precision, recall, f1, _ = precision_recall_fscore_support(
            all_labels, all_preds, average="weighted", zero_division=0
        )
        return acc, precision, recall, f1

    fold_accuracies, fold_precisions, fold_recalls, fold_f1s, fold_results = [], [], [], [], []
    student_fold_accuracies, student_fold_precisions, student_fold_recalls, student_fold_f1s = [], [], [], []
    # Track QAT fold metrics
    qat_fold_accuracies, qat_fold_precisions, qat_fold_recalls, qat_fold_f1s = [], [], [], []

    completed_folds_path = os.path.join(output_dir, "completed_folds.txt")
    completed_folds = []
    if os.path.exists(completed_folds_path):
        with open(completed_folds_path, 'r') as f:
            completed_folds = [int(x.strip()) for x in f.readlines() if x.strip()]
        print(f"Resuming — folds already done: {completed_folds}")

    for fold, (train_idx, val_idx) in enumerate(skf.split(np.arange(len(dataset)), targets)):
        print(f"\n========== Fold {fold+1}/5 ==========")

        if (fold + 1) in completed_folds:
            print(f"Fold {fold+1} already completed — skipping.")
            fold_csv = os.path.join(output_dir, f"fold{fold+1}_results.json")
            if os.path.exists(fold_csv):
                with open(fold_csv, 'r') as f:
                    r = json.load(f)
                fold_accuracies.append(r['teacher_acc'])
                fold_precisions.append(r['teacher_prec'])
                fold_recalls.append(r['teacher_rec'])
                fold_f1s.append(r['teacher_f1'])
                student_fold_accuracies.append(r['student_acc'])
                student_fold_precisions.append(r['student_prec'])
                student_fold_recalls.append(r['student_rec'])
                student_fold_f1s.append(r['student_f1'])
                qat_fold_accuracies.append(r['qat_acc'])
                qat_fold_precisions.append(r['qat_prec'])
                qat_fold_recalls.append(r['qat_rec'])
                qat_fold_f1s.append(r['qat_f1'])
                fold_results.append(r)
            continue

        train_data = SafeImageFolder(data_dir, transform=train_transform)
        val_data   = SafeImageFolder(data_dir, transform=val_transform)

        train_subset = Subset(train_data, train_idx)
        val_subset   = Subset(val_data, val_idx)

        train_loader = DataLoader(train_subset, batch_size=batch_size, shuffle=True, num_workers=4)
        val_loader   = DataLoader(val_subset, batch_size=batch_size, shuffle=False, num_workers=4)

        # New model each fold
        model = timm.create_model("vit_base_patch16_224", pretrained=True, num_classes=num_classes).to(device)

        param_groups = get_layer_wise_params(model, base_lr=1e-4, decay=0.65)
        optimizer = torch.optim.AdamW(param_groups, weight_decay=1e-4)

        train_labels_fold = [dataset.samples[i][1] for i in train_idx]
        class_counts_fold = torch.zeros(num_classes)
        for lbl in train_labels_fold:
            class_counts_fold[lbl] += 1
        class_weights = len(train_labels_fold) / (num_classes * class_counts_fold.clamp(min=1))
        class_weights = class_weights.to(device)
        criterion = nn.CrossEntropyLoss(label_smoothing=0.1, weight=class_weights)

        from torch.optim.lr_scheduler import CosineAnnealingLR, LinearLR, SequentialLR
        warmup_scheduler = LinearLR(optimizer, start_factor=0.1, end_factor=1.0, total_iters=5)
        cosine_scheduler = CosineAnnealingLR(optimizer, T_max=num_epochs - 5, eta_min=1e-6)
        scheduler = SequentialLR(optimizer, schedulers=[warmup_scheduler, cosine_scheduler], milestones=[5])

        early_stopping = EarlyStopping(patience=15, delta=0.001, verbose=True,
                                       path=os.path.join(output_dir, f"best_fold{fold+1}.pt"))

        variant = f"fold{fold+1}"
        train_model(model, train_loader, val_loader, criterion, optimizer,
                    scheduler, num_epochs, device, early_stopping, variant, output_dir)

        # --- After teacher is trained for this fold ---
        print(f"\nStarting Student Training for Fold {fold+1}")

        # Reload best teacher model for this fold
        teacher_model = timm.create_model("vit_base_patch16_224", pretrained=False, num_classes=num_classes)
        teacher_model.load_state_dict(torch.load(os.path.join(output_dir, f"best_fold{fold+1}.pt")))
        teacher_model.to(device)
        teacher_model.eval()

        # Initialize student
        student_model = StudentNet(num_classes=num_classes).to(device)

        criterion = nn.CrossEntropyLoss(label_smoothing=0.1, weight=class_weights)
        optimizer = torch.optim.AdamW(student_model.parameters(), lr=2e-4, weight_decay=1e-4)
        scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=num_epochs, eta_min=1e-6)
        early_stopping = EarlyStopping(patience=12, delta=0.001, verbose=True,
                                      path=os.path.join(output_dir, f"best_student_fold{fold+1}.pt"))

        variant = f"student_fold{fold+1}"
        student_model, epoch_df, metrics = train_student_model(
            student_model, teacher_model, train_loader, val_loader,
            criterion, optimizer, scheduler,
            num_epochs=num_epochs, device=device,
            early_stopping=early_stopping,
            variant=variant, output_dir=output_dir
        )

        plot_metrics(*metrics, variant=f"student_fold{fold+1}", output_dir=output_dir)

        student_ckpt_path = os.path.join(output_dir, f"best_student_fold{fold+1}.pt")

        print(f"\nSaving clean student model for Fold {fold+1}...")
        save_clean_student(
            student_model,
            save_path_fp32=os.path.join(output_dir, f"clean_student_fp32_fold{fold+1}.pt"),
            save_path_fp16=None,
            num_classes=num_classes,
            device=device
        )

        # --- Train QAT student (fine-tune KD student under FP16) ---
        qat_model, qat_model_fp16, qat_ckpt_path, fp16_path, qat_metrics = train_qat_student_model(
        student_ckpt_path=student_ckpt_path,
        teacher_model=teacher_model,
        train_loader=train_loader,
        val_loader=val_loader,
        device=device,
        variant=f"qat_student_fold{fold+1}",
        output_dir=output_dir
    )

        # Unpack metrics from QAT training
        train_losses, val_losses, train_accs, val_accs = qat_metrics

        plot_qat_metrics(train_losses, val_losses, train_accs, val_accs,
                        variant=f"qat_student_fold{fold+1}",
                        output_dir=output_dir)


        # --- Evaluate teacher ---
        teacher_acc, teacher_prec, teacher_rec, teacher_f1 = evaluate_on_loader(model, val_loader, device)

        fold_accuracies.append(teacher_acc)
        fold_precisions.append(teacher_prec)
        fold_recalls.append(teacher_rec)
        fold_f1s.append(teacher_f1)

        # Evaluate student
        student_acc, student_prec, student_rec, student_f1 = evaluate_on_loader(student_model, val_loader, device)
        print(f"Student Fold {fold+1} - Acc: {student_acc:.4f}, Precision: {student_prec:.4f}, Recall: {student_rec:.4f}, F1: {student_f1:.4f}")

        student_fold_accuracies.append(student_acc)
        student_fold_precisions.append(student_prec)
        student_fold_recalls.append(student_rec)
        student_fold_f1s.append(student_f1)

        # --- Evaluate QAT student ---
        qat_preds, qat_targets = [], []
        qat_model_fp16.eval()
        with torch.no_grad():
            for images, labels in val_loader:
                valid_mask = labels != -1
                if not valid_mask.any(): continue
                images, labels = images[valid_mask].to(device), labels[valid_mask].to(device)
                outputs = qat_model_fp16(images.half())
                preds = outputs.argmax(1)
                qat_preds.extend(preds.cpu().numpy())
                qat_targets.extend(labels.cpu().numpy())


        qat_acc = accuracy_score(qat_targets, qat_preds) * 100
        qat_prec, qat_rec, qat_f1, _ = precision_recall_fscore_support(qat_targets, qat_preds, average="weighted", zero_division=0)
        print(f"[Fold {fold+1}] QAT Acc: {qat_acc:.2f}% | Prec: {qat_prec:.3f} | Rec: {qat_rec:.3f} | F1: {qat_f1:.3f}")

        qat_fold_accuracies.append(qat_acc/100)
        qat_fold_precisions.append(qat_prec)
        qat_fold_recalls.append(qat_rec)
        qat_fold_f1s.append(qat_f1)

        fold_results.append({
            "fold": fold+1,
            "teacher_acc": teacher_acc, "teacher_prec": teacher_prec, "teacher_rec": teacher_rec, "teacher_f1": teacher_f1,
            "student_acc": student_acc, "student_prec": student_prec, "student_rec": student_rec, "student_f1": student_f1,
            "qat_acc": qat_acc/100, "qat_prec": qat_prec, "qat_rec": qat_rec, "qat_f1": qat_f1,
        })

        fold_result_to_save = {
            "fold": fold + 1,
            "teacher_acc": teacher_acc, "teacher_prec": teacher_prec,
            "teacher_rec": teacher_rec, "teacher_f1": teacher_f1,
            "student_acc": student_acc, "student_prec": student_prec,
            "student_rec": student_rec, "student_f1": student_f1,
            "qat_acc": float(qat_acc/100), "qat_prec": float(qat_prec),
            "qat_rec": float(qat_rec), "qat_f1": float(qat_f1),
        }
        with open(os.path.join(output_dir, f"fold{fold+1}_results.json"), 'w') as fj:
            json.dump(fold_result_to_save, fj)
        with open(completed_folds_path, 'a') as fc:
            fc.write(f"{fold+1}\n")
        print(f"Fold {fold+1} marked complete — results saved.")

    print("\n===== 5-Fold Cross Validation Results (Teacher) =====")
    print(f"Accuracy:  {np.mean(fold_accuracies):.4f} ± {np.std(fold_accuracies):.4f}")
    print(f"Precision: {np.mean(fold_precisions):.4f} ± {np.std(fold_precisions):.4f}")
    print(f"Recall:    {np.mean(fold_recalls):.4f} ± {np.std(fold_recalls):.4f}")
    print(f"F1 Score:  {np.mean(fold_f1s):.4f} ± {np.std(fold_f1s):.4f}")

    print("\n===== 5-Fold Cross Validation Results (Student) =====")
    print(f"Accuracy:  {np.mean(student_fold_accuracies):.4f} ± {np.std(student_fold_accuracies):.4f}")
    print(f"Precision: {np.mean(student_fold_precisions):.4f} ± {np.std(student_fold_precisions):.4f}")
    print(f"Recall:    {np.mean(student_fold_recalls):.4f} ± {np.std(student_fold_recalls):.4f}")
    print(f"F1 Score:  {np.mean(student_fold_f1s):.4f} ± {np.std(student_fold_f1s):.4f}")

    print("\n===== 5-Fold Cross Validation Results (QAT Student) =====")
    print(f"Accuracy:  {np.mean(qat_fold_accuracies):.4f} ± {np.std(qat_fold_accuracies):.4f}")
    print(f"Precision: {np.mean(qat_fold_precisions):.4f} ± {np.std(qat_fold_precisions):.4f}")
    print(f"Recall:    {np.mean(qat_fold_recalls):.4f} ± {np.std(qat_fold_recalls):.4f}")
    print(f"F1 Score:  {np.mean(qat_fold_f1s):.4f} ± {np.std(qat_fold_f1s):.4f}")

    # Save per-fold results to CSV
    results_df = pd.DataFrame({
        "Fold": list(range(1, len(fold_accuracies) + 1)),
        "Teacher Accuracy": fold_accuracies,
        "Teacher Precision": fold_precisions,
        "Teacher Recall": fold_recalls,
        "Teacher F1": fold_f1s,
        "Student Accuracy": student_fold_accuracies,
        "Student Precision": student_fold_precisions,
        "Student Recall": student_fold_recalls,
        "Student F1": student_fold_f1s,
        "QAT Accuracy": qat_fold_accuracies,
        "QAT Precision": qat_fold_precisions,
        "QAT Recall": qat_fold_recalls,
        "QAT F1": qat_fold_f1s,
    })

    csv_path = os.path.join(output_dir, "crossval_results_teacher_student_qat.csv")
    results_df.to_csv(csv_path, index=False)
    print(f"\nPer-fold results saved to: {csv_path}")
    print(results_df)

    # ======================================================
    # Final Clean Result Output (FP32 vs FP16 QAT Student)
    # ======================================================

    # Compute efficiency metrics for student model
    def count_params(model):
        return sum(p.numel() for p in model.parameters() if p.requires_grad)

    def file_size_mb(path):
        return os.path.getsize(path) / (1024 * 1024)

    fp32_params_total = count_params(student_model.cpu())
    fp32_params_deployed = sum(p.numel() for n, p in student_model.named_parameters()
                               if p.requires_grad and not n.startswith('feature_adapter'))
    fp32_est_mb_total    = fp32_params_total    * 4 / 1024 / 1024
    fp32_est_mb_deployed = fp32_params_deployed * 4 / 1024 / 1024

    # QAT model size
    fp16_size_mb_total = file_size_mb(fp16_path) if os.path.exists(fp16_path) else float("nan")
    # FP16 deployed size estimate
    adapter_params = fp32_params_total - fp32_params_deployed
    fp16_size_mb_deployed = (fp32_params_deployed * 2) / 1024 / 1024

    # flops = 413.9e6

    def measure_inference_latency(model, loader, n_warmup=10, n_runs=50, use_half=False):
        import time
        model.eval()
        model = model.to(device)
        if use_half:
            model = model.half()
        images, labels = next(iter(loader))
        images = images.to(device)
        if use_half:
            images = images.half()
        # Warmup
        for _ in range(n_warmup):
            _ = model(images)
        torch.cuda.synchronize()
        start = time.time()
        for _ in range(n_runs):
            _ = model(images)
        torch.cuda.synchronize()
        end = time.time()
        return (end - start) / n_runs / images.size(0) * 1000

    latency_fp32 = measure_inference_latency(student_model, val_loader)
    latency_fp16 = measure_inference_latency(qat_model_fp16, val_loader, use_half=True)

    teacher_acc = np.mean(fold_accuracies) * 100
    teacher_prec = np.mean(fold_precisions)
    teacher_rec = np.mean(fold_recalls)
    teacher_f1 = np.mean(fold_f1s)

    student_acc = np.mean(student_fold_accuracies) * 100
    student_prec = np.mean(student_fold_precisions)
    student_rec = np.mean(student_fold_recalls)
    student_f1 = np.mean(student_fold_f1s)

    qat_acc = np.mean(qat_fold_accuracies) * 100
    qat_prec = np.mean(qat_fold_precisions)
    qat_rec = np.mean(qat_fold_recalls)
    qat_f1 = np.mean(qat_fold_f1s)

    print("\n================= Final Results =================")
    print(f"{'Model Variant':15} | {'Accuracy':8} | {'Precision':9} | {'Recall':7} | {'F1-score':8} | {'Params':7} | {'Size (MB)':9} | {'FLOPs':8} | {'Latency (ms/img)':17}")
    print("-"*110)
    print(f"{'Teacher (ViT-B16)':15} | {teacher_acc:.2f}%   | {teacher_prec:.3f}     | {teacher_rec:.3f}   | {teacher_f1:.3f}    | {'-':7} | {'-':9} | {'-':8} | {'-':17}")
    print(f"{'FP32 (KD Student)':15} | {student_acc:.2f}%   | {student_prec:.3f}     | {student_rec:.3f}   | {student_f1:.3f}    | {fp32_params_deployed/1e6:.2f}M | {fp32_est_mb_deployed:.2f} MB  | {flops/1e6:.1f}M | {latency_fp32:.2f}")
    print(f"{'FP16 (QAT Student)':15} | {qat_acc:.2f}%   | {qat_prec:.3f}     | {qat_rec:.3f}   | {qat_f1:.3f}    | {fp32_params_deployed/1e6:.2f}M | {fp16_size_mb_deployed:.2f} MB  | {flops/1e6:.1f}M | {latency_fp16:.2f}")
    print("="*110)
    print(f"\nNote — Checkpoint file sizes (include training-only feature_adapter: +{(fp32_params_total-fp32_params_deployed)/1e6:.2f}M params):")
    print(f"  FP32 checkpoint : {fp32_est_mb_total:.2f} MB  ({fp32_params_total/1e6:.2f}M params total)")
    print(f"  FP16 checkpoint : {fp16_size_mb_total:.2f} MB  ({fp32_params_total/1e6:.2f}M params total)")
    print(f"  Deployed FP32   : {fp32_est_mb_deployed:.2f} MB  ({fp32_params_deployed/1e6:.2f}M params — feature_adapter excluded)")
    print(f"  Deployed FP16   : {fp16_size_mb_deployed:.2f} MB  ({fp32_params_deployed/1e6:.2f}M params — feature_adapter excluded)")

    # ======================================================
    # Visualization: Teacher vs Student vs QAT (Mean ± SD)
    # ======================================================
    metrics = ["Accuracy", "Precision", "Recall", "F1"]

    teacher_means = [
        np.mean(fold_accuracies),
        np.mean(fold_precisions),
        np.mean(fold_recalls),
        np.mean(fold_f1s),
    ]
    teacher_stds = [
        np.std(fold_accuracies),
        np.std(fold_precisions),
        np.std(fold_recalls),
        np.std(fold_f1s),
    ]

    student_means = [
        np.mean(student_fold_accuracies),
        np.mean(student_fold_precisions),
        np.mean(student_fold_recalls),
        np.mean(student_fold_f1s),
    ]
    student_stds = [
        np.std(student_fold_accuracies),
        np.std(student_fold_precisions),
        np.std(student_fold_recalls),
        np.std(student_fold_f1s),
    ]

    qat_means = [
        np.mean(qat_fold_accuracies),
        np.mean(qat_fold_precisions),
        np.mean(qat_fold_recalls),
        np.mean(qat_fold_f1s),
    ]
    qat_stds = [
        np.std(qat_fold_accuracies),
        np.std(qat_fold_precisions),
        np.std(qat_fold_recalls),
        np.std(qat_fold_f1s),
    ]

    # Make per-fold dataframe from fold_results
    results_df = pd.DataFrame(fold_results)

    print("\n================ Per-Fold Results ================")
    print(results_df.to_string(index=False))

    print("\n================= Mean ± Std =================")
    print(f"Teacher: Acc {np.mean(fold_accuracies):.4f} ± {np.std(fold_accuracies):.4f}, F1 {np.mean(fold_f1s):.4f} ± {np.std(fold_f1s):.4f}")
    print(f"Student: Acc {np.mean(student_fold_accuracies):.4f} ± {np.std(student_fold_accuracies):.4f}, F1 {np.mean(student_fold_f1s):.4f} ± {np.std(student_fold_f1s):.4f}")
    print(f"QAT:     Acc {np.mean(qat_fold_accuracies):.4f} ± {np.std(qat_fold_accuracies):.4f}, F1 {np.mean(qat_fold_f1s):.4f} ± {np.std(qat_fold_f1s):.4f}")
    print("=================================================")


    # Plot grouped bar chart with error bars
    x = np.arange(len(metrics))
    width = 0.25

    plt.figure(figsize=(10, 6))
    plt.bar(x - width, teacher_means, width, yerr=teacher_stds, label="Teacher", capsize=5)
    plt.bar(x, student_means, width, yerr=student_stds, label="Student", capsize=5)
    plt.bar(x + width, qat_means, width, yerr=qat_stds, label="QAT Student", capsize=5)

    plt.xticks(x, metrics)
    plt.ylabel("Score")
    plt.ylim(0, 1.0)
    plt.title("Cross-Validation Performance (Mean ± SD)")
    plt.legend()
    plt.grid(axis="y", linestyle="--", alpha=0.6)

    plot_path = os.path.join(output_dir, "crossval_teacher_student_qat_comparison.png")
    plt.savefig(plot_path, dpi=300, bbox_inches="tight")
    plt.show()

    print(f"\nComparison plot saved to: {plot_path}")

if __name__ == "__main__":
    import sys

    if 'ipykernel' in sys.modules:
        class Args:
            epochs = 100
            batch_size = 32
            learning_rate = 2e-5
            data_dir = data_dir
        args = Args()
    else:
        parser = argparse.ArgumentParser(description="Train Vision Transformer for potato leaf disease classification")
        parser.add_argument('--epochs', type=int, default=100)
        parser.add_argument('--batch_size', type=int, default=32)
        parser.add_argument('--learning_rate', type=float, default=2e-5)
        parser.add_argument('--data_dir', type=str, required=True)
        args = parser.parse_args()

    main(args)


```
